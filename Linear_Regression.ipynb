```python
df = df.drop(['car_name' , 'model'], axis = 1)
```

```python
df = pd.get_dummies(df , columns=['brand', 'seller_type', 'fuel_type', 'transmission_type'] , drop_first=True)
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression , Ridge , Lasso
```

```python
# Training and Testing Phase - Train and Test and Split

x = df.drop(columns=['selling_price'])
y = df['selling_price']

X_train , X_test, y_train, y_test = train_test_split(x, y , test_size=0.2 , random_state=42)

print(f"Train Set:- {X_train.shape[0]}")
print(f"Test set :- {X_test.shape[0]}")
```

```python
X_train.describe().round(1)
```

```python
df
```

```python
# Model Training Phase

numerical_cols = ['vehicle_age' ,'km_driven' , 'mileage','engine','max_power','seats']

scaler = StandardScaler()

X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])

X_train
```

```python
model = LinearRegression()

model.fit(X_train , y_train)

y_train_pred = model.predict(X_train)

y_train_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae_train = mean_absolute_error(y_train , y_train_pred)

mae_train
```

```python
y_test_pred = model.predict(X_test)
```

```python

mae_test = mean_absolute_error(y_test , y_test_pred)
mae_test
```

```python
# Calculating the baseline

baseline_mean = df['selling_price'].mean().round(2)

baseline = abs(df['selling_price'] - baseline_mean)

mae = sum(baseline) / len(baseline)
mae_2 = baseline.mean()
mae
```

```python
train_r2 = r2_score(y_train, y_train_pred)
train_r2    
```

```python
test_r2 = r2_score(y_test, y_test_pred)
test_r2
```

```python
print(f"Trained R2_score :- {train_r2:.4f}")
print(f"Tested R2_score :- {test_r2:.4f}")
```

```python
# Doing Ridge and Lasso Regressions

model_rid = Ridge(alpha = 1.0)

model_rid.fit(X_train , y_train)

y_train_rid = model_rid.predict(X_train)
y_test_rid = model_rid.predict(X_test)

rid_tr_mae = mean_absolute_error(y_train , y_train_rid)
rid_te_mae = mean_absolute_error(y_test , y_test_rid)

rid_tr_r2 = r2_score(y_train , y_train_rid)
rid_te_r2 = r2_score(y_test , y_test_rid)

print(f"{"Train":<15}  {"Test":>15}")
print(f"{rid_tr_mae:<15.4f}  {rid_te_mae:>15.4f}")
print(f"{rid_tr_r2:<15.4f}  {rid_tr_r2:>15.4f}")


```
