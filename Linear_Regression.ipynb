```python
import pandas as pd 

df = pd.read_csv("cardekho_dataset.csv")

df
```

```python
df.info()
```

```python
# Dropping the unwanted colum in the dataset

df = df.drop(["Unnamed: 0"] , axis=1)
```

```python
df
```

```python
df.columns
```

```python
brand_counts = df["brand"].value_counts() 
```

```python
rare_brands = brand_counts[brand_counts < 200].index.tolist()

df['brand'] = df['brand'].replace(rare_brands , "Others")

```

```python
df = df.drop(['car_name' , 'model'], axis = 1)
```

```python
df
```

```python
df = pd.get_dummies(df , columns=['brand', 'seller_type', 'fuel_type', 'transmission_type'] , drop_first=True)
```
